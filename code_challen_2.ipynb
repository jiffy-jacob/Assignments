```python
import math
```

```python
#creating class employee
class Employee:
  def __init__(self,name,basic_salary):
    self.name=name
    self.basic_salary=basic_salary

  def calculate_hra(self):
    return 0.20*self.basic_salary   #20% of salary

  def calculate_da(self):
    return 0.10*self.basic_salary   #10% of salary

  def calculate_totalsalary(self):
    hra=self.calculate_hra()
    totalsalary=self.basic_salary + hra + self.calculate_da()
    return totalsalary
```

```python
#creating objects
emp1=Employee("Aman",30000)
emp2=Employee("Rahul",50000)
```

```python
#display total salary
print("Total salary of",emp1.name,"is",emp1.calculate_totalsalary())
print("Total salary of",emp2.name,"is",emp2.calculate_totalsalary())
```
